## Advanced analytics

Focus: List16 coverage and CGPA by high school via transcript join.

```python
from analysis_utils import basic_profile, missingness_report

print(basic_profile(df))
missingness_report(df, top_n=20)
```

```python
# Join to List16 transcript for CGPA by high school
trans = pd.read_csv(DATA_DIR / "student_transcript_list16.csv")
trans["CGPA"] = pd.to_numeric(trans["CGPA"], errors="coerce")

merged = df.merge(trans[["REG_NO","CGPA"]], on="REG_NO", how="left")

merged.groupby("HIGH_SCHOOL")["CGPA"].agg(["count","mean","median","std"]).sort_values("mean", ascending=False).head(30)
```
